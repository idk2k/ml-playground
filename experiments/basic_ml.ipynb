```python
v = np.array([3.0, 4.0])
w = np.array([5.0, 12.0])

# method a:
alpha = np.arctan2(v[1], v[0])
beta = np.arctan2(w[1], w[0])
alpha, beta
```

```python
cos_alpha, sin_alpha = v[0] / np.linalg.norm(v), v[1] / np.linalg.norm(v)
cos_beta, sin_beta = w[0] / np.linalg.norm(w), w[1] / np.linalg.norm(w)

cos_theta_trig = cos_alpha * cos_beta + sin_alpha * sin_beta

# 3. Способ Б: Через скалярное произведение (как в векторизованном ML)
cos_theta_vector = np.dot(v, w) / (np.linalg.norm(v) * np.linalg.norm(w))
print(f"Cos(theta) через тригонометрию: {cos_theta_trig:.4f}")
print(f"Cos(theta) через векторную формулу: {cos_theta_vector:.4f}")
print(f"Результаты совпадают: {np.isclose(cos_theta_trig, cos_theta_vector)}")
```
